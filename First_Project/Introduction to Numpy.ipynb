## Dot product Example (nut butter sales)

```python
np.random.seed(0)
# Number of jars sold
sales_amount = np. random.randint(20, size=(5,3))
sales_amount
```

```python
weekly_sales = pd.DataFrame(sales_amount, index=["Mon", "Tues", "Wed", "Thurs", "Fri"], columns=["Almund butter", "Peanut butter", "Cashew butter"])
weekly_sales
```

```python
# Create prices array

prices = np.array([10, 8, 12])
prices
```

```python
# Create butter_prices DataFrame
butter_prices = pd.DataFrame(prices.reshape(1, 3), index=["Price"], columns=["Almund butter", "Peanut butter", "Cashew butter"])
butter_prices
```

```python
total_sales = prices.dot(sales_amount.T)
total_sales
```

```python
# Create daily_sales
butter_prices
```

```python
daily_sales = butter_prices.dot(weekly_sales.T)
daily_sales
```

```python
weekly_sales
```

```python
weekly_sales["Total ($)"] = daily_sales.T
weekly_sales
```
