```python
# Creating Column from Series
seats_column = pd.Series([5, 5, 5, 5, 5])

# New column called seats
car_sales["Seats"] = seats_column
car_sales
```

```python
# car_sales.fillna({"Seats":5}, inplace=True)
car_sales["Seats"] = car_sales["Seats"].fillna(5)
car_sales
```

```python
# Creating Column from Python List
fuel_economy = [7.5, 8.6, 6.3, 7.7, 9.2, 2.4, 5.3, 8.6, 6.8, 4.1]
car_sales["Fuel per 100KM"] = fuel_economy
car_sales
```

```python
car_sales["Total fuel used "] = car_sales["Odometer (KM)"]/100 * car_sales["Fuel per 100KM"]
```

```python
#Creating a Column form another Column
car_sales["Total fuel used (L)"] = car_sales["Odometer (KM)"]/100 * car_sales["Fuel per 100KM"]
car_sales
```

```python
#Creating a Column from a Single Vallue
car_sales["Number of Wheels"] = 4
car_sales
```

```python
car_sales["Passed road Safety"] = True
car_sales
```

```python
car_sales = car_sales.drop("Total fuel used ", axis=1)
car_sales
```

```python
car_sales_shufflled = car_sales.sample(frac=1)
car_sales_shufflled
```
